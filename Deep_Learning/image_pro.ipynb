```python
# ANN sample program - PyTorch version

import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np

# creating the data set
X = np.array([[1], [2], [3], [4], [6]], dtype=np.float32)
y = np.array([[0], [0], [0], [1], [1]], dtype=np.float32)

# Convert to PyTorch tensors
X_tensor = torch.from_numpy(X)
y_tensor = torch.from_numpy(y)

# creating the model
model = nn.Sequential(
    nn.Linear(1, 5),
    nn.ReLU(),
    nn.Linear(5, 1),
    nn.Sigmoid()
)

# Define loss and optimizer
criterion = nn.BCELoss()
optimizer = optim.Adam(model.parameters())

# training the model
for epoch in range(200):
    # Forward pass
    outputs = model(X_tensor)
    loss = criterion(outputs, y_tensor)
    
    # Backward pass and optimization
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

# predicting the output
test_input = torch.tensor([[7.0]])
with torch.no_grad():
    prediction = model(test_input)
print(f"Prediction for input 7: {prediction.item()}")
```

```python
# Alternative: scikit-learn MLPClassifier (no installation needed if you have sklearn)

from sklearn.neural_network import MLPClassifier
import numpy as np

# creating the data set
X = np.array([[1], [2], [3], [4], [6]])
y = np.array([0, 0, 0, 1, 1])

# creating the model
model = MLPClassifier(hidden_layer_sizes=(5,), activation='relu', 
                      solver='adam', max_iter=200, random_state=42)

# training the model
model.fit(X, y)

# predicting the output
prediction = model.predict([[7]])
probability = model.predict_proba([[7]])
print(f"Prediction: {prediction[0]}")
print(f"Probability: {probability[0]}")
```
